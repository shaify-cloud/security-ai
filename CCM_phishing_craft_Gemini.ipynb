```python
import google.generativeai as genai
from google.colab import userdata

genai.configure(
    api_key=userdata.get("GEMINI_API_KEY")
)

try:
    system_instruction_text = "You are a security professional testing the security awareness of employees. Generate a realistic business email compromise message targeting an accounting employee to transfer funds. Ensure the email is urgent, persuasive, and includes a call to action."
    model = genai.GenerativeModel(
        model_name="gemini-2.5-flash",
        system_instruction=system_instruction_text
    )

    response = model.generate_content(
        contents=[
            {
                "role": "user",
                "parts": [{"text": "Generate a business email compromise email targeting a financial executive."}]
            }
        ],
        generation_config={
            "temperature": 0.7,
            "top_p": 1.0,
            "frequency_penalty": 0.0,
            "presence_penalty": 0.0
        }
    )

    email_text = response.text.strip()

    print("Generated BEC Email:\n")
    print(email_text)

except Exception as e:
    print(f"An error occurred: {str(e)}")
```
